```python
import pandas as pd
import glob
import matplotlib.pyplot as plt 
file_list = glob.glob('txt/*.txt')
data = []

for file in file_list:
    with open(file, 'r') as f:
        lines = f.readlines()[100:]
        # extract the first column of data
        col_data = [float(line.split()[1]) for line in lines if float(line.split()[1]) != 0]
        data.append(col_data)
fig, ax = plt.subplots()
ax.boxplot(data)
ax.set_xticklabels(file_list)
ax.set_ylabel('CPU Utilisation')

plt.show()
```

```python
import pandas as pd
import glob
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np
file_list = glob.glob('txt/*.txt')
data = []

for file in file_list:
    with open(file, 'r') as f:
        lines = f.readlines()[100:]
        # extract the first column of data
        col_data = [float(line.split()[3]) for line in lines if float(line.split()[3]) != 0]
        data.append(col_data)
# create the boxplot using seaborn and add grid lines
sns.boxplot(data=data)

# get the current axis object
ax = plt.gca()

# set y-axis limits to include all data
ax.set_ylim([min([min(x) for x in data]), max([max(x) for x in data])])

ax.grid(axis='y', alpha=1.0)
# set x-axis tick labels and y-axis label

# add minor gridlines
ax.yaxis.grid(True, which='minor', linestyle='--', alpha=1.0)

plt.xticks(np.arange(len(file_list)), file_list)
plt.ylabel('Value')

plt.show()
```

```python
import pandas as pd
import glob
import matplotlib.pyplot as plt 
import re
file_list = glob.glob('txt/*.txt')
data = []

for file in file_list:
    with open(file, 'r') as f:
        lines = f.readlines()[100:]
        # extract the first column of data
        col_data = [float(line.split()[1]) for line in lines if float(line.split()[1]) > 10]
        data.append(col_data)
fig, ax = plt.subplots()
ax.boxplot(data, notch=True)
ax.minorticks_on()
xnames = []
for file in file_list:
    tempsplit=re.split('[-\.]',file)
    print(tempsplit)
    xnames.append(tempsplit[1]+tempsplit[2])
print(xnames)
ax.tick_params(axis='x', labelsize=8)
ax.set_xticklabels(xnames)
ax.set_ylabel('CPU Utilization % (on Raspberry Pi 3B+)')
# add major gridlines
ax.grid(axis='y', alpha=1.0, linestyle='-')

# add minor gridlines
ax.yaxis.grid(True, which='minor', linestyle='-', alpha=1.0, linewidth=0.2)

plt.show()
```
